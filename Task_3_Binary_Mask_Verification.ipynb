```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

TRAIN_IMG = "/content/drive/MyDrive/coco20k/train/images"
TRAIN_MASK = "/content/drive/MyDrive/coco20k/train/masks"

files = sorted(os.listdir(TRAIN_IMG))[:20]

plt.figure(figsize=(12, 50))
idx = 1

for f in files:
    img = cv2.imread(os.path.join(TRAIN_IMG, f))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    mask_path = os.path.join(TRAIN_MASK, f.replace(".jpg", ".png"))
    mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)

    mask_bin = (mask > 0).astype(np.uint8) * 255

    plt.subplot(20, 3, idx); idx += 1
    plt.imshow(img); plt.axis("off")

    plt.subplot(20, 3, idx); idx += 1
    plt.imshow(mask, cmap="gray"); plt.axis("off")

    plt.subplot(20, 3, idx); idx += 1
    plt.imshow(mask_bin, cmap="gray"); plt.axis("off")

plt.tight_layout()
plt.savefig("/content/task3_mask_verification.png")
plt.show()
```
