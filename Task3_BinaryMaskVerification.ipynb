```python
import os, cv2, numpy as np, matplotlib.pyplot as plt

IMG_DIR = "/content/drive/MyDrive/coco20k/images"
MASK_DIR = "/content/drive/MyDrive/coco20k/masks"

files = sorted(os.listdir(IMG_DIR))[:20]

plt.figure(figsize=(12, 50))
idx = 1

for f in files:
    img_path = os.path.join(IMG_DIR, f)
    mask_path = os.path.join(MASK_DIR, f.replace(".jpg", ".png"))

    img = cv2.cvtColor(cv2.imread(img_path), cv2.COLOR_BGR2RGB)
    mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)

    mask_bin = (mask > 0).astype(np.uint8)

    plt.subplot(20, 2, idx); idx += 1
    plt.imshow(img); plt.axis("off")

    plt.subplot(20, 2, idx); idx += 1
    plt.imshow(mask_bin, cmap="gray"); plt.axis("off")

plt.tight_layout()
plt.savefig("/content/task3_output_20pairs.png")
plt.show()
```
